```python
fig1, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 6))

# Rugplot without hue (you'll have to plot separately for each 'time')
sns.rugplot(data=df, ax=ax1, y='tip', x='total_bill', height=0.1)
sns.scatterplot(data=df, ax=ax1, y='tip', x='total_bill', hue='time')

sns.rugplot(data=df, ax=ax2, y='tip', x='size', height=0.1)
sns.scatterplot(data=df, ax=ax2, y='tip', x='size', hue='time')

plt.show()
```

```python
#rugplot with scatter plot ( hue parameter

fig1, ax1 = plt.subplots()

sns.rugplot(data=df, ax=ax1, y='tip', x='total_bill', height=0.15)
sns.scatterplot(data=df, ax=ax1, y='tip', x='total_bill', hue='time')

```
